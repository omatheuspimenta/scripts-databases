```python
def standardize_time(val):
    if pd.isna(val):
        return val
    
    # Basic cleaning: remove escaped commas, replace underscores with spaces, and strip whitespace
    s = str(val).replace(r'\,', ',').replace('_', ' ').strip()
    
    # If the string is purely numeric, return it as is (the measure unit is not specified, so we leave it unchanged)
    if re.fullmatch(r'\d+', s):
        return s

    s_lower = s.lower()

    # Convert spelled-out numbers to digits (e.g., "one" -> "1", "two" -> "2", "one and a half" -> "1.5")
    word_to_num = {
        'one and a half': '1.5',
        'one': '1', 'two': '2', 'three': '3', 'four': '4',
        'five': '5', 'six': '6', 'seven': '7', 'eight': '8',
        'nine': '9', 'ten': '10'
    }
    for word, digit in word_to_num.items():
        s_lower = re.sub(rf'\b{word}\b', digit, s_lower)

    # Acronyms and abbreviations for time points in biological contexts
    # Example: 120daf, daf120, 120-daf, 90 dafb -> 120 DAF
    m_daf = re.search(r'(?:daf[b]?\s*(\d+)|\b(\d+)\s*[-_]?\s*daf[b]?\b)', s_lower)
    if m_daf:
        num = m_daf.group(1) or m_daf.group(2)
        return f"{num} DAF"

    # Ex: 120dpa, 120 dpa-1 -> 120 DPA
    m_dpa = re.search(r'\b(\d+)\s*[-_]?\s*dpa(?:\s*[-_]?\s*\d+)?\b', s_lower)
    if m_dpa:
        return f"{m_dpa.group(1)} DPA"

    # Ex: 0 dai, 29 dai (day after somatic embryogenesis induction) -> X DAI
    m_dai = re.search(r'\b(\d+)\s*[-_]?\s*dai\b', s_lower)
    if m_dai:
        return f"{m_dai.group(1)} DAI"

    # Ex: 1 dpi, 14 dpi -> X DPI
    m_dpi = re.search(r'\b(\d+)\s*[-_]?\s*dpi\b', s_lower)
    if m_dpi:
        return f"{m_dpi.group(1)} DPI"

    # Ex: 48 wpg, 12 wpg -> X WPG
    m_wpg = re.search(r'\b(\d+)\s*[-_]?\s*wpg\b', s_lower)
    if m_wpg:
        return f"{m_wpg.group(1)} WPG"

    # Ex: 7 daa (days after anthesis) -> 7 DAA
    m_daa = re.search(r'\b(\d+)\s*[-_]?\s*daa\b', s_lower)
    if m_daa:
        return f"{m_daa.group(1)} DAA"

    # Ex: 3 dba (days before anthesis) -> 3 DBA
    m_dba = re.search(r'\b(\d+)\s*[-_]?\s*dba\b', s_lower)
    if m_dba:
        return f"{m_dba.group(1)} DBA"

    # Standardize time points
    # Days after flowering / bloom / blossoming -> DAF
    m_after_flow = re.search(r'(\d+)\s*(?:days?|d)\s*(?:after|post)\s*(?:the\s*)?(?:flowering|flower|full bloom|bloom|blossoming)', s_lower)
    if m_after_flow:
        return f"{m_after_flow.group(1)} DAF"

    # Days before flowering / anthesis -> DBA
    m_before_flow = re.search(r'(\d+)\s*(?:days?|d)\s*(?:before|prior to)\s*(?:the\s*)?(?:flowering|flower|anthesis)', s_lower)
    if m_before_flow:
        return f"{m_before_flow.group(1)} DBA"

    # Days after pollination -> DAP
    m_after_poll = re.search(r'(\d+)\s*(?:days?|d)\s*(?:after|post)\s*pollin(?:ation|eng)', s_lower)
    if m_after_poll:
        return f"{m_after_poll.group(1)} DAP"

    # Days after treatment -> DAT
    m_after_treat = re.search(r'(\d+)\s*(?:days?|d)\s*(?:after|post)\s*treatment', s_lower)
    if m_after_treat:
        return f"{m_after_treat.group(1)} DAT"

    # Weeks post-exposure / post-inoculation -> WPI
    m_wpi = re.search(r'(\d+)\s*(?:weeks?|w)\s*(?:post[- ]exposure|post[- ]inoculation)', s_lower)
    if m_wpi:
        return f"{m_wpi.group(1)} WPI"

    # Days after petal fall (ex: "Stage 2 (176 Days After Petal Fall, Colour Break)")
    m_dapf = re.search(r'(\d+)\s*days?\s*after\s*petal\s*fall', s_lower)
    if m_dapf:
        return f"{m_dapf.group(1)} DAPF"

    # Explicit time units (days, weeks, months, years) with optional ranges or suffixes
    # Years (ex: "1-2 years old plant", "5+ years", "3 years", "12 year")
    m_year_range = re.search(r'(\d+\s*[-–]\s*\d+)\s*years?', s_lower)
    if m_year_range:
        return f"{m_year_range.group(1).replace(' ', '')} years"
    
    m_year = re.search(r'(\d+(?:\.\d+)?)\s*(?:\+)?\s*(?:years?|yeas?)\b', s_lower)
    if m_year:
        return f"{m_year.group(1)} years"

    # Months (ex: "4-month-old", "2 month old", "12 m", "6 months")
    m_month = re.search(r'(\d+)\s*(?:months?|month-old|m)\b', s_lower)
    if m_month and not re.search(r'\d+\s*cm', s_lower):  # evita casar "cm"
        return f"{m_month.group(1)} months"

    # Weeks (ex: "8-week-old", "18 weeks", "2 w", "1 w")
    m_week = re.search(r'(\d+)\s*(?:weeks?|week-old|w)\b', s_lower)
    if m_week:
        return f"{m_week.group(1)} weeks"

    # Days (ex: "45 days", "21days", "90d", "0d")
    m_day = re.search(r'\b(\d+)\s*(?:days?|d)\b', s_lower)
    if m_day:
        return f"{m_day.group(1)} days"

    # Fallback
    return s
```

### Isolation Source

```python
# Check unique values and count in the 'isolation_source' column
df["isolation_source"] = df["isolation_source"].str.strip()
isolation_source_counts = df["isolation_source"].value_counts()
print(isolation_source_counts)
```

```python
def standardize_isolation_source(val):
    if pd.isna(val):
        return val
    
    # basic cleaning: remove escaped commas, replace underscores with spaces, and strip whitespace
    s = str(val).replace(r'\,', ',').replace('_', ' ').strip()
    s_clean = " ".join(s.split())
    s_lower = s_clean.lower()

    # flower and reproductive tissues
    if any(k in s_lower for k in ['pollen mother', 'microsporocyte', 'microspore', 'tetrad']):
        return 'microspore/pollen mother cell'
    if 'stamen primord' in s_lower:
        return 'stamen primordia'
    if 'ovule' in s_lower:
        return 'ovule'
    if 'style' in s_lower:
        return 'style'
    if any(k in s_lower for k in ['flower', 'fiower', 'anthesis', 'bloom']):
        return 'flower'
    if 'bud' in s_lower:
        return 'bud'

    # fruit tissues
    if any(k in s_lower for k in ['subepidermis', 'sub-epidermis']):
        return 'fruit subepidermis'
    if 'epidermis' in s_lower:
        return 'fruit epidermis'
    if any(k in s_lower for k in ['peel', 'pericarp']):
        return 'fruit peel'
    if any(k in s_lower for k in ['juice sac', 'pulp']):
        return 'fruit juice sac'
    if 'seed' in s_lower and 'seedling' not in s_lower:
        return 'seed'

    # fruit
    if any(k in s_lower for k in ['fruit', 'fruitlet', 'color break', 'breaker', 'colored', 
                                  'full ripe', 'ripening', 'mature green']):
        return 'fruit'

    # in vitro tissues
    if any(k in s_lower for k in ['callus', 'calli']):
        return 'callus'
    if 'protoplast' in s_lower:
        return 'protoplast'

    # vegetative tissues
    if any(k in s_lower for k in ['leaf', 'leaves']):
        return 'leaf'
    if any(k in s_lower for k in ['stem', 'internode']):
        return 'stem'
    if 'root' in s_lower:
        return 'root'
    if 'seedling' in s_lower:
        return 'seedling'
    if 'tree' in s_lower:
        return 'whole plant/tree'

    # Resolve species-level identification based on keywords in the isolation source
    # Citrus sinensis (Sweet Orange)
    if re.search(r'\b(citrus\s+sinensis|sinensis|sweet\s+orange|orange|osbeck|wanjincheng|newhall|cara\s+cara)\b', s_lower):
        return 'Citrus sinensis'

    # Citrus limon (Lemon)
    if re.search(r'\b(citrus\s+limon|limon|lemon)\b', s_lower) and 'persian' not in s_lower:
        return 'Citrus limon'

    # Citrus maxima (Pummelo / Pomelo)
    if re.search(r'\b(citrus\s+maxima|maxima|grandis|pomelo|pummelo)\b', s_lower):
        return 'Citrus maxima'

    # Citrus reticulata (Mandarin / Tangerine)
    if re.search(r'\b(citrus\s+reticulata|reticulata|mandarin|tangerine|shatangju|ponkan)\b', s_lower):
        return 'Citrus reticulata'

    # Citrus paradisi (Grapefruit)
    if re.search(r'\b(grapefruit|paradisi)\b', s_lower):
        return 'Citrus paradisi'

    # Other Citrus species and hybrids
    if re.search(r'\b(citrus\s+medica|xiangyuan)\b', s_lower):
        return 'Citrus medica'
    if 'clementina' in s_lower:
        return 'Citrus clementina'
    if 'junos' in s_lower or s_lower == 'citj':
        return 'Citrus junos'
    if 'fortunella' in s_lower or 'kumquat' in s_lower:
        return 'Fortunella spp.'
    if 'persian lime' in s_lower:
        return 'Citrus latifolia'
    if 'flying dragon' in s_lower or 'poncirus' in s_lower or 'trifoliata' in s_lower:
        return 'Poncirus trifoliata'
    if 'cleopatra' in s_lower:
        return 'Citrus reshni (Cleopatra mandarin)'
    if 'us 897' in s_lower or 'us-897' in s_lower:
        return 'Citrus hybrid (US-897)'

    # Citrus spp. generic identification (if only "citrus" is present, optionally followed by a number)
    if re.fullmatch(r'citrus(?:\s+\d+)?', s_lower):
        return 'Citrus spp.'

    # replicate identifiers and numbers at the end of the string
    # Ex: "Wt S2 Replicate 3" -> "WT S2", "Mt S3 Replicate 1" -> "MT S3"
    s_clean = re.sub(r'\s*\b(?:biological\s+)?replicate\s*[-_]?\s*\d+\b', '', s_clean, flags=re.IGNORECASE)
    s_clean = re.sub(r'\s*\b(?:rep|rp)\s*[-_]?\s*\d+\b', '', s_clean, flags=re.IGNORECASE)

    # Ex: "Ck R1" -> "CK", "D Np R2" -> "D NP", "D R3" -> "D"
    s_clean = re.sub(r'\s*\b[rR]\d+\b$', '', s_clean)

    # Ex: "Gx Pt Phase4 3" -> "Gx Pt Phase4", "T78 1" -> "T78", "Dvs 2" -> "Dvs"
    # Remove trailing numbers after PhaseN or after a word of 2-10 letters
    s_clean = re.sub(r'(?<=\bPhase\d)\s+\d+$', '', s_clean, flags=re.IGNORECASE)
    s_clean = re.sub(r'(?<=\b[A-Za-z]{2,10})\s+\d+$', '', s_clean)

    return s_clean.strip()
```

### Treatment

```python
# Check unique values and count in the 'treatment' column
df["treatment"] = df["treatment"].str.strip()
treatment_counts = df["treatment"].value_counts()
print(treatment_counts)
```

```python
def standardize_treatment(val):
    if pd.isna(val):
        return val
    
    # Basic cleaning: remove escaped commas, replace underscores with spaces, and strip whitespace
    s = str(val).replace(r'\,', ',').replace('_', ' ').strip()
    s = " ".join(s.split())
    s_lower = s.lower()

    # Controls and untreated samples
    if re.fullmatch(r'(control|ck|ctr|normal|untreated|check|healthy control|healthy \(control\))(?:\s*[-_]?\s*\d+)?', s_lower):
        return 'Control'
    if re.fullmatch(r'healty\s*\d+', s_lower):
        return 'Healthy control'
    if re.fullmatch(r'(wildtype|wild\s*type|wild|wt)(?:\s*[-_]?\s*\d+)?(?:\s*plant)?', s_lower):
        return 'Wild-type (control)'
    if re.fullmatch(r'mock(?:\s*[-_]?\s*\d+)?', s_lower):
        return 'Mock'
    if s_lower in ['no', 'no interstock', 'without film packaging']:
        return s.title()
    if s_lower == 'with interstock':
        return 'With interstock'

    # temporal controls and time points
    if re.fullmatch(r'control\s*baseline', s_lower):
        return 'Control (baseline)'
    m_ctrl_wk = re.match(r'control\s*(?:week|week1)?\s*(\d+)', s_lower)
    if m_ctrl_wk:
        return f"Control (Week {m_ctrl_wk.group(1)})"
    if re.match(r'ck\s*0w', s_lower):
        return 'Control (0 weeks)'
    if re.match(r'ck\s*([48])\s*wap', s_lower):
        m = re.search(r'ck\s*([48])\s*wap', s_lower)
        return f"Control ({m.group(1)} WAP)"
    if re.match(r'control\s*0\s*[-_]?\s*\d*', s_lower) or s_lower in ['0d', '0d 1', '0d 2', '0d 3']:
        return 'Control (0 d)'
    if re.match(r'control\s*48\s*[-_]?\s*\d*', s_lower):
        return 'Control (48 h)'
    if re.match(r'treatment\s*48\s*[-_]?\s*\d*', s_lower):
        return 'Treatment (48 h)'
    if 'control, no extra light' in s_lower:
        return 'Control (no extra light)'

    # abiotic stress treatments
    # salt stress (NaCl)
    if s_lower in ['salt', 'nacl treatment']:
        return 'Salt stress (NaCl)'
    if re.match(r's\s*([12])\s*w', s_lower):
        m = re.search(r's\s*([12])\s*w', s_lower)
        return f"Salt stress ({m.group(1)} week)"

    # Drought stress and recovery
    if s_lower in ['drought', 'drought stress', 'drought-stress']:
        return 'Drought stress'
    if s_lower == 'non drought stress':
        return 'Non-drought stress (well-watered control)'
    if 'drought with bamboo powder bio-fertilizer' in s_lower:
        return 'Drought stress + bamboo powder bio-fertilizer'
    if 'drought stress with arbuscular mycorrhizal fungi' in s_lower:
        return 'Drought stress + AMF'
    if 'recover from drought' in s_lower:
        return 'Drought recovery'
    if s_lower == 'snp drought':
        return 'Drought stress + SNP (nitric oxide)'
    if s_lower == 'snp control':
        return 'Control + SNP (nitric oxide)'

    # cold stress and PEG-mediated osmotic stress
    if s_lower == 'cold stress':
        return 'Cold stress'
    if s_lower == 'peg':
        return 'PEG-mediated osmotic stress'

    # light treatments and photoperiods
    m_light = re.match(r'white[- ]light\s*5000\s*lx\s*,\s*(\d+)\s*h', s_lower)
    if m_light:
        return f"White light (5000 lx, {m_light.group(1)} h)"
    if s_lower in ['blue light', 'red light', 'white light', 'dark']:
        return s.title()

    # nutritional treatments and deficiencies
    if 'mg-deficiency' in s_lower or 'mg deficient' in s_lower:
        return 'Mg deficiency (0 mM Mg(NO3)2)'
    if 'mg-sufficiency' in s_lower or 'mg sufficient' in s_lower:
        return 'Mg sufficiency (2 mM Mg(NO3)2)'
    if re.match(r'n0\s*[-_]?\s*\d*', s_lower):
        return 'Nitrogen deficiency (0 mM N)'
    if re.match(r'n15\s*[-_]?\s*\d*', s_lower):
        return 'Nitrogen sufficiency (15 mM N)'

    # storage conditions and durations
    m_store_rt = re.search(r'store\s*(\d+)\s*days\s*under\s*room\s*temperature', s_lower)
    if m_store_rt:
        return f"Storage at room temperature ({m_store_rt.group(1)} days)"
    
    # storage at different temperatures: Ex: "15-7d-1", "25-14d-2", "35-21d-3"
    m_temp_storage = re.match(r'(\d+)\s*[-_]\s*(\d+)\s*d(?:\s*[-_]\s*\d+)?', s_lower)
    if m_temp_storage:
        return f"Storage at {m_temp_storage.group(1)}°C ({m_temp_storage.group(2)} days)"

    # Kumquat storage conditions with or without film packaging
    if 'kumquats refrigerated for 50 days using film packaging' in s_lower:
        return 'Refrigerated storage (50 days, with film packaging)'
    if 'kumquats refrigerated for 50 days without film packaging' in s_lower:
        return 'Refrigerated storage (50 days, without film packaging)'
    if s_lower in ['fresh kumquats', 'fresh-harvest']:
        return 'Fresh harvest (baseline)'

    # chemical treatments and combinations
    # Cu + coumarin: Ex: "Cu0.5Cou100_2" -> "0.5 µM Cu + 100 µM Coumarin"
    m_cucou = re.match(r'cu\s*([0-9.]+)\s*cou\s*([0-9.]+)', s_lower)
    if m_cucou:
        return f"Cu {m_cucou.group(1)} µM + Coumarin {m_cucou.group(2)} µM"

    # Cu + HA: Ex: "Lcu400Ha0.5_2", "Rcu0.5Ha0_1"
    m_cuha = re.match(r'([lr])cu\s*([0-9.]+)\s*ha\s*([0-9.]+)', s_lower)
    if m_cuha:
        tissue = 'Leaf' if m_cuha.group(1) == 'l' else 'Root'
        return f"{tissue}: Cu {m_cuha.group(2)} µM + HA {m_cuha.group(3)} mg/L"

    # pH and combinations with Cu and Al: Ex: "R-Ph3+Cu-2", "Ph4-Al-1", "Ph4.8-2"
    m_ph = re.match(r'(r[- ])?ph\s*([0-9.]+)(?:\s*([+|-])\s*(cu|al))?', s_lower)
    if m_ph:
        prefix = "Root: " if m_ph.group(1) else ""
        ph_val = f"pH {m_ph.group(2)}"
        stress = f" + {m_ph.group(4).upper()}" if m_ph.group(4) else ""
        return f"{prefix}{ph_val}{stress}".strip()

    # cu toxicity and cu400 treatments
    if re.match(r'cu[- ]?toxicity', s_lower) or re.match(r'cu400', s_lower):
        return 'Copper toxicity (400 µM Cu)'
    if s_lower in ['0mm al', 'aluminum stress']:
        return 'Aluminum stress' if '0mm' not in s_lower else '0 mM Al (control)'
    if s_lower == '4mm al':
        return 'Aluminum stress (4 mM Al)'

    # pathogen infections and inoculations
    # HLB / Ca. Liberibacter asiaticus infection and healthy controls
    m_hlb_h = re.match(r'hlb\s*(\d+)\s*h', s_lower)
    if m_hlb_h:
        return f"HLB / CLas infection ({m_hlb_h.group(1)} h)"
    m_h_h = re.match(r'h\s*(\d+)\s*h', s_lower)
    if m_h_h:
        return f"Healthy control ({m_h_h.group(1)} h)"
    m_hlb_wk = re.match(r'hlb\s*week\s*(\d+)', s_lower)
    if m_hlb_wk:
        return f"HLB / CLas infection (Week {m_hlb_wk.group(1)})"
    if re.match(r'clas\s*([48])\s*wap', s_lower):
        m = re.search(r'clas\s*([48])\s*wap', s_lower)
        return f"CLas inoculation ({m.group(1)} WAP)"
    if re.match(r'clas\s*inoculated\s*(\d+)\s*wpi', s_lower):
        m = re.search(r'clas\s*inoculated\s*(\d+)\s*wpi', s_lower)
        return f"CLas inoculation ({m.group(1)} WPI)"
    if re.match(r'mock[- ]inoculated\s*(\d+)\s*wpi', s_lower):
        m = re.search(r'mock[- ]inoculated\s*(\d+)\s*wpi', s_lower)
        return f"Mock inoculation ({m.group(1)} WPI)"

    # Coinfections and other pathogens
    if re.match(r'b2\s*b6', s_lower):
        return 'CTV (B2 + B6 coinfection)'
    if re.match(r'b2\s*clas', s_lower):
        return 'CTV (B2) + CLas coinfection'
    if re.match(r'b6\s*clas', s_lower):
        return 'CTV (B6) + CLas coinfection'
    if re.match(r'cevd', s_lower):
        return 'CEVd infection'
    if re.match(r'cbcvd', s_lower):
        return 'CBCVd infection'
    if s_lower == 'cyvcv':
        return 'CYVCV infection'

    # field samples and natural infections
    if 'calas-infected citrus tree' in s_lower or 'infected with huanglongbing' in s_lower:
        return 'CLas-infected tree'
    if 'healthy cirtus tree' in s_lower or 'healthy citrus medica' in s_lower:
        return 'Healthy control tree'

    # salicylic acid / BABA
    m_sa = re.match(r'(\d+)\s*h\s*after\s*2\s*mm\s*sa\s*treatment', s_lower)
    if m_sa:
        return f"2 mM Salicylic acid treatment ({m_sa.group(1)} h)"
    if '20 mm beta-aminobutyric acid' in s_lower:
        return '20 mM BABA treatment'

    # overexpression and transgenic lines
    # overexpression in citrus peel: Ex: "Citrus peel transiently overexpressing CsWRKY22-1" -> "Transient overexpression: CsWRKY22 in fruit peel"
    m_trans_peel = re.search(r'citrus peel transiently overexpressing\s+([a-z0-9-_ ]+?)(?:-\d+)?$', s_lower)
    if m_trans_peel:
        gene = m_trans_peel.group(1).strip().title()
        return f"Transient overexpression: {gene} in fruit peel"

    # RpfF / Wild-Type vs CLas
    if 'rpff transgenic plant infected' in s_lower:
        return 'RpfF-transgenic + CLas infected'
    if 'rpff transgenic plant non-infected' in s_lower:
        return 'RpfF-transgenic + uninfected control'
    if 'wild-type plant (non-transgenic) infected' in s_lower:
        return 'Wild-type + CLas infected'
    if 'wild-type plant (non-transgenic) non-infected' in s_lower:
        return 'Wild-type + uninfected control'

    # Lemon WT vs TG
    if 'transgenic lemon' in s_lower:
        return 'Transgenic lemon'
    if 'wild-type lemon' in s_lower:
        return 'Wild-type lemon'

    # callus / calli overexpression lines
    if 'crwrky42-overexpression calli' in s_lower:
        return 'CrWRKY42-overexpression callus'
    if 'wild type calli' in s_lower:
        return 'Wild-type callus'
    m_ri_line = re.search(r'the\s*([a-z0-9-]+)\s*line\s*at\s*15\s*dai', s_lower)
    if m_ri_line:
        return f"{m_ri_line.group(1).upper()} line (15 DAI)"

    # AtNPR1 e FT
    if 'atnpr1-transgenic' in s_lower:
        return 'AtNPR1-transgenic'
    if 'ft-transgenic' in s_lower:
        return 'FT-transgenic'
    if 'non-transgenic' in s_lower:
        return 'Non-transgenic control'
    m_zc = re.search(r'transgenic\s*(zc\d+)', s_lower)
    if m_zc:
        return f"Transgenic {m_zc.group(1).upper()}"

    # sufixes and replicate identifiers
    # Remove "Biological Replicate(s) X", "Repeat-X", "Replicate X", etc.
    s_clean = re.sub(r'[,;]?\s*\b(?:biological\s+)?(?:replicates?|repeats?|rep)\s*[-_]?\s*\d+\b', '', s, flags=re.IGNORECASE)
    # Remove numeric suffixes (ex: "Clas 1" -> "CLas", "Hini_2" -> "Hini")
    s_clean = re.sub(r'[-_ ]\b\d+\b$', '', s_clean)

    return s_clean.strip()
```

```python
def map_time_to_ontology(value: str) -> str | float:
    """
    Maps a given time-related value to its corresponding ontology terms from Plant Ontology (PO), Plant Experimental Conditions Ontology (PECO), and Units of measurement ontology (UO).
    The function checks for matches in the provided dictionaries and returns a comma-separated string of matched ontology terms. If no matches are found, it returns NaN.
    
    Args:
        value (str): The time-related value to be mapped to ontology terms.
    Returns:
        str | float: A comma-separated string of matched ontology terms or NaN if no matches    
    """
    if pd.isna(value):
        return np.nan

    value_clean = str(value).strip()
    matched_terms = []

    # PO - Plant Ontology (developmental stages)
    po_term = time2PO.get(value_clean)
    if po_term and po_term not in matched_terms:
        matched_terms.append(po_term)

    # PECO - Plant Experimental Conditions Ontology (specific treatments/events)
    for k, peco_term in plant_experimental2PECO.items():
        if peco_term:
            # Check for whole word matches to avoid partial matches (e.g., "DAI" should not match "Dairy")
            if re.search(r"\b" + re.escape(k) + r"\b", value_clean):
                if peco_term not in matched_terms:
                    matched_terms.append(peco_term)
                    break

    # UO - Units of measurement ontology (UO) - 
    # Sort the keys by length in descending order to prioritize longer matches (e.g., "weeks" before "week")
    for k, uo_term in sorted(
        time2UO.items(), key=lambda x: len(x[0]), reverse=True
    ):
        if uo_term:
            if re.search(
                r"\b" + re.escape(k) + r"\b", value_clean, re.IGNORECASE
            ):
                if uo_term not in matched_terms:
                    matched_terms.append(uo_term)
                    break

    if not matched_terms:
        return np.nan

    return ", ".join(matched_terms)
```

```python
df["time_ontology"] = df["time"].apply(map_time_to_ontology)
```
